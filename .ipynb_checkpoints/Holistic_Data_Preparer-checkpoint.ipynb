## Part A: Conceptual Foundation

**1. What is Data Analysis?**
Data Analysis = raw data ko clean, explore aur interpret karke usse meaningful insights nikalna, jisse better decisions liye ja sakein.
*Project mein:* raw customer records analyse karke pata chalega kaunse patterns default risk se juday hain (e.g., low credit score + missed payments → high default chance).

**2. How to Plan a Data Science Project?**
Har DS project **CRISP-DM** framework follow karta hai — 6 phases: Business Understanding → Data Understanding → Data Preparation → Modeling → Evaluation → Deployment.
*Project mein:* hum sirf **Data Understanding + Data Preparation** phase kar rahe hain — data ko ML-ready banana hai.

**3. How to Frame a Machine Learning Problem?**
Vague business need ko precise **Input → Output → Metric** mein convert karna.

| | Detail |
|---|---|
| Input | age, income, credit_score, repayment_history, transaction_count, spending_ratio, etc. |
| Output | `default_flag` → 0 (No Default) / 1 (Default) |
| Task type | Binary Classification |
| Metric | Recall on defaulters matters more than raw accuracy (business risk) |

**4. Tensors — In-Depth Explanation**
Tensor = numbers ka N-dimensional array/container. Ek customer ka single score = 0D, ek customer ki saari features = 1D, poora dataset = 2D, agar time bhi add ho (monthly history) = 3D+.

```python
# ===== Tensors demo (NumPy) =====
import numpy as np

credit_score_0d = np.array(750)                          # 0D - scalar
customer_1d = np.array([35, 650000, 750, 2, 34])          # 1D - vector (one customer's features)
customers_2d = np.array([                                 # 2D - matrix (multiple customers x features)
    [35, 650000, 750, 2, 34],
    [42, 480000, 690, 1, 55],
    [29, 900000, 610, 4, 12]
])
customers_3d = np.stack([customers_2d, customers_2d + 5])  # 3D - tensor (customers x time x features)

print("0D shape:", credit_score_0d.shape, "| ndim:", credit_score_0d.ndim)
print("1D shape:", customer_1d.shape, "| ndim:", customer_1d.ndim)
print("2D shape:", customers_2d.shape, "| ndim:", customers_2d.ndim)
print("3D shape:", customers_3d.shape, "| ndim:", customers_3d.ndim)
```

*Interpretation:* Jab humara `main_transactions.csv` pandas mein load hoga aur model ko diya jaayega, wo internally ek **2D tensor** ban jaata hai — rows = customers, columns = features.

## Part B: Data Acquisition


```python
# ===== Imports =====
import pandas as pd
import numpy as np
import json
import sqlite3
import matplotlib.pyplot as plt
from sklearn.impute import SimpleImputer, KNNImputer, MissingIndicator
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
from sklearn.preprocessing import (
    OrdinalEncoder, LabelEncoder, OneHotEncoder,
    StandardScaler, MinMaxScaler, MaxAbsScaler, RobustScaler, Normalizer,
    FunctionTransformer, PowerTransformer, KBinsDiscretizer
)
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.cluster import KMeans

pd.set_option("display.max_columns", None)
```

```python
# ===== 1. Load CSV - main transactions dataset =====
df_csv = pd.read_csv("D:\Holistic-Data-Preparer-\data_set\main_transactions.csv")
print(df_csv.shape)
df_csv.head()
```

```python
# ===== 2. Load JSON - customer metadata =====
with open("customer_metadata.json") as f:
    metadata_records = json.load(f)
df_json = pd.DataFrame(metadata_records)
print(df_json.shape)
df_json.head()
```

```python
# ===== 3. Load SQL - loan repayment history =====
conn = sqlite3.connect("repayment_history.db")
df_sql = pd.read_sql("SELECT * FROM loan_repayments", conn)
conn.close()
print(df_sql.shape)
df_sql.head()
```

```python
# ===== 4. Fetch from a dummy API - external economic indicators =====
def fetch_economic_indicators_api(region):
    """Mock API call - simulates an external economic-indicators endpoint."""
    mock_db = {
        "North": {"inflation_rate": 5.4, "unemployment_rate": 6.1},
        "South": {"inflation_rate": 4.8, "unemployment_rate": 5.3},
        "East":  {"inflation_rate": 5.9, "unemployment_rate": 7.2},
        "West":  {"inflation_rate": 5.1, "unemployment_rate": 5.8},
    }
    return mock_db[region]

df_api = pd.DataFrame(df_json["region"].apply(fetch_economic_indicators_api).tolist())
df_api["customer_id"] = df_json["customer_id"]
df_api = df_api[["customer_id", "inflation_rate", "unemployment_rate"]]
df_api.head()
```

```python
# ===== Merge all 4 sources on customer_id =====
df_raw = df_csv.merge(df_json, on="customer_id").merge(df_sql, on="customer_id").merge(df_api, on="customer_id")
print("Merged shape:", df_raw.shape)
df_raw.head()
```

## Part C: Data Understanding & Cleaning

```python
# ===== Explore dataset structure =====
df_raw.info()
```

```python
# ===== Explore dataset statistics =====
df_raw.describe(include="all").T
```

```python
# ===== Manual data quality report (Pandas Profiling alternative) =====
def data_quality_report(df):
    report = pd.DataFrame({
        "dtype": df.dtypes,
        "missing_count": df.isnull().sum(),
        "missing_%": (df.isnull().mean() * 100).round(2),
        "n_unique": df.nunique(),
    })
    return report

quality_report = data_quality_report(df_raw)
print("Duplicate rows:", df_raw.duplicated().sum())
quality_report
```

### Missing Value Handling


```python
# ===== Simple Imputer - numerical (age: mean/median) =====
num_imputer = SimpleImputer(strategy="median")
age_imputed = num_imputer.fit_transform(df_raw[["age"]])
print("Missing before:", df_raw["age"].isnull().sum(), "| after:", pd.isnull(age_imputed).sum())
```

```python
# ===== Simple Imputer - categorical (employment_type: most frequent) =====
cat_imputer = SimpleImputer(strategy="most_frequent")
employment_imputed = cat_imputer.fit_transform(df_raw[["employment_type"]])
print("Missing before:", df_raw["employment_type"].isnull().sum(), "| after:", pd.isnull(employment_imputed).sum())
```

```python
# ===== Most Frequent Category Imputation (manual, on gender) =====
most_frequent_gender = df_raw["gender"].mode()[0]
gender_imputed = df_raw["gender"].fillna(most_frequent_gender)
print("Most frequent category:", most_frequent_gender)
print("Missing before:", df_raw["gender"].isnull().sum(), "| after:", gender_imputed.isnull().sum())
```

```python
# ===== Missing Indicator + Random Sample Imputation (annual_income) =====
missing_indicator = MissingIndicator()
income_missing_flag = missing_indicator.fit_transform(df_raw[["annual_income"]])
df_temp = df_raw.copy()
df_temp["annual_income_was_missing"] = income_missing_flag.astype(int)

non_null_samples = df_temp["annual_income"].dropna()
np.random.seed(42)
random_fill = np.random.choice(non_null_samples, size=df_temp["annual_income"].isnull().sum())
income_random_imputed = df_temp["annual_income"].copy()
income_random_imputed[income_random_imputed.isnull()] = random_fill
print("Flag column preview:\n", df_temp["annual_income_was_missing"].value_counts())
```

```python
# ===== KNN Imputer (annual_income, loan_amount, credit_score) =====
knn_cols = ["annual_income", "loan_amount", "credit_score"]
knn_imputer = KNNImputer(n_neighbors=5)
knn_imputed = pd.DataFrame(knn_imputer.fit_transform(df_raw[knn_cols]), columns=knn_cols)
print(knn_imputed.isnull().sum())
knn_imputed.head()
```

```python
# ===== MICE Algorithm (Iterative Imputer) on same columns =====
mice_imputer = IterativeImputer(random_state=42, max_iter=10)
mice_imputed = pd.DataFrame(mice_imputer.fit_transform(df_raw[knn_cols]), columns=knn_cols)
print(mice_imputed.isnull().sum())
mice_imputed.head()
```

```python
# ===== Complete Case Analysis (demo only - not used for final data) =====
complete_case_df = df_raw.dropna()
print("Original rows:", len(df_raw), "| After dropping all missing rows:", len(complete_case_df))
print(f"Data loss: {(1 - len(complete_case_df)/len(df_raw))*100:.1f}%")
```

```python
# ===== Final Imputation Strategy - build df_imputed =====
df_imputed = df_raw.copy()

# categorical -> most frequent
df_imputed["gender"] = df_imputed["gender"].fillna(df_imputed["gender"].mode()[0])
df_imputed["employment_type"] = cat_imputer.fit_transform(df_imputed[["employment_type"]]).ravel()

# age -> median
df_imputed["age"] = num_imputer.fit_transform(df_imputed[["age"]]).ravel()

# correlated financial columns -> KNN Imputer
df_imputed[knn_cols] = knn_imputer.fit_transform(df_imputed[knn_cols])

print("Remaining missing values:\n", df_imputed.isnull().sum().sum())
```

---
## Part D: Outlier Handling


```python
# ===== Visualize outliers (boxplots) =====
fig, axes = plt.subplots(1, 3, figsize=(12, 4))
for ax, col in zip(axes, knn_cols):
    ax.boxplot(df_imputed[col])
    ax.set_title(col)
plt.tight_layout()
plt.show()
```

```python
# ===== Z-score Method (credit_score) =====
from scipy import stats
z_scores = np.abs(stats.zscore(df_imputed["credit_score"]))
z_outliers = df_imputed[z_scores > 3]
print("Z-score outliers found:", len(z_outliers))
z_outliers[["customer_id", "credit_score"]]
```

```python
# ===== IQR Method (loan_amount) =====
Q1 = df_imputed["loan_amount"].quantile(0.25)
Q3 = df_imputed["loan_amount"].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
iqr_outliers = df_imputed[(df_imputed["loan_amount"] < lower_bound) | (df_imputed["loan_amount"] > upper_bound)]
print(f"IQR bounds: [{lower_bound:.0f}, {upper_bound:.0f}] | Outliers found: {len(iqr_outliers)}")
```

```python
# ===== Percentile Method (annual_income) =====
lower_pct = df_imputed["annual_income"].quantile(0.01)
upper_pct = df_imputed["annual_income"].quantile(0.99)
pct_outliers = df_imputed[(df_imputed["annual_income"] < lower_pct) | (df_imputed["annual_income"] > upper_pct)]
print(f"1st-99th percentile bounds: [{lower_pct:.0f}, {upper_pct:.0f}] | Outliers found: {len(pct_outliers)}")
```

```python
# ===== Winsorization Technique (annual_income) =====
from scipy.stats.mstats import winsorize
income_winsorized = winsorize(df_imputed["annual_income"], limits=[0.01, 0.01])
print("Before max:", df_imputed["annual_income"].max(), "| After winsorize max:", income_winsorized.max())
```

```python
# ===== Apply chosen outlier treatment -> df_treated =====
df_treated = df_imputed.copy()

# credit_score: clip to valid domain range (300-850) - domain-driven fix for erroneous entries
df_treated["credit_score"] = df_treated["credit_score"].clip(300, 850)

# annual_income & loan_amount: IQR-based capping (keeps row count, fixes skew-driving outliers)
for col in ["annual_income", "loan_amount"]:
    Q1, Q3 = df_treated[col].quantile([0.25, 0.75])
    IQR = Q3 - Q1
    lb, ub = Q1 - 1.5 * IQR, Q3 + 1.5 * IQR
    df_treated[col] = df_treated[col].clip(lb, ub)

print(df_treated[knn_cols].describe())
```

## Part E: Feature Engineering

```python
# ===== Mixed Variables (concept demo) =====
mixed_example = pd.Series(["2", "3", "5+", "Unknown", "10"])
mixed_numeric = pd.to_numeric(mixed_example, errors="coerce")   # numeric part
mixed_category = mixed_example.where(mixed_numeric.isnull())    # categorical/text part
print("Numeric part:\n", mixed_numeric.tolist())
print("Category part:\n", mixed_category.dropna().tolist())
```

```python
# ===== Date & Time - extract Year, Month, Day, Weekday from join_date =====
df_treated["join_date"] = pd.to_datetime(df_treated["join_date"])
df_treated["join_year"] = df_treated["join_date"].dt.year
df_treated["join_month"] = df_treated["join_date"].dt.month
df_treated["join_day"] = df_treated["join_date"].dt.day
df_treated["join_weekday"] = df_treated["join_date"].dt.day_name()

df_treated[["join_date", "join_year", "join_month", "join_day", "join_weekday"]].head()
```

### Encoding Categorical Variables

```python
# ===== Ordinal Encoding (education_level - has natural order) =====
edu_order = [["Primary", "Secondary", "Graduate", "Post-Graduate"]]
ordinal_encoder = OrdinalEncoder(categories=edu_order)
df_treated["education_level_encoded"] = ordinal_encoder.fit_transform(df_treated[["education_level"]])
df_treated[["education_level", "education_level_encoded"]].drop_duplicates().sort_values("education_level_encoded")
```

```python
# ===== Label Encoding (gender - binary-ish feature) =====
label_encoder = LabelEncoder()
df_treated["gender_encoded"] = label_encoder.fit_transform(df_treated["gender"])
dict(zip(label_encoder.classes_, label_encoder.transform(label_encoder.classes_)))
```

```python
# ===== One-Hot Encoding (region, loan_purpose - no natural order) =====
df_treated = pd.get_dummies(df_treated, columns=["region", "loan_purpose"], prefix=["region", "purpose"])
[c for c in df_treated.columns if c.startswith("region_") or c.startswith("purpose_")]
```

### Encoding Numerical Features

```python
# ===== Binning (discretize annual_income into groups) =====
df_treated["income_band"] = pd.cut(
    df_treated["annual_income"], bins=4,
    labels=["Low", "Medium", "High", "Very High"]
)
df_treated["income_band"].value_counts()
```

```python
# ===== Binarization (flag credit_score > 700) =====
df_treated["good_credit_flag"] = (df_treated["credit_score"] > 700).astype(int)
df_treated["good_credit_flag"].value_counts()
```

```python
# ===== Quantile Binning (transaction_count - equal-sized groups) =====
df_treated["txn_quantile_bin"] = pd.qcut(df_treated["transaction_count"], q=4, labels=["Q1", "Q2", "Q3", "Q4"], duplicates="drop")
df_treated["txn_quantile_bin"].value_counts()
```

```python
# ===== K-Means Binning (transaction_count - data-driven clusters) =====
kmeans_binner = KMeans(n_clusters=3, random_state=42, n_init=10)
df_treated["txn_kmeans_bin"] = kmeans_binner.fit_predict(df_treated[["transaction_count"]])
df_treated.groupby("txn_kmeans_bin")["transaction_count"].agg(["min", "max", "count"])
```

# ---
## Part F: Feature Scaling


```python
# ===== Standardization (Z-score scaling) =====
scale_cols = ["annual_income", "loan_amount"]
standard_scaler = StandardScaler()
standardized = standard_scaler.fit_transform(df_treated[scale_cols])
print("Mean ~0:", standardized.mean(axis=0).round(2), "| Std ~1:", standardized.std(axis=0).round(2))
```

```python
# ===== Normalization (row-wise, unit norm) =====
normalizer = Normalizer()
normalized = normalizer.fit_transform(df_treated[scale_cols])
print("Row vector norms (should be ~1):", np.linalg.norm(normalized, axis=1)[:5].round(2))
```

```python
# ===== Min-Max Scaling =====
minmax_scaler = MinMaxScaler()
minmax_scaled = minmax_scaler.fit_transform(df_treated[scale_cols])
print("Min:", minmax_scaled.min(axis=0), "| Max:", minmax_scaled.max(axis=0))
```

```python
# ===== MaxAbs Scaling =====
maxabs_scaler = MaxAbsScaler()
maxabs_scaled = maxabs_scaler.fit_transform(df_treated[scale_cols])
print("Max abs value (should be 1):", np.abs(maxabs_scaled).max(axis=0))
```

```python
# ===== Robust Scaling (uses median/IQR - outlier resistant) =====
robust_scaler = RobustScaler()
robust_scaled = robust_scaler.fit_transform(df_treated[scale_cols])
print("Median ~0:", np.median(robust_scaled, axis=0).round(2))
```

---
## Part G: Feature Construction & Transformation

```python
# ===== PowerTransformer - Box-Cox (loan_amount) & Yeo-Johnson (annual_income) =====
boxcox_transformer = PowerTransformer(method="box-cox")   # needs strictly positive values
loan_boxcox = boxcox_transformer.fit_transform(df_treated[["loan_amount"]])

yeojohnson_transformer = PowerTransformer(method="yeo-johnson")  # works with zero/negative too
income_yeojohnson = yeojohnson_transformer.fit_transform(df_treated[["annual_income"]])

print("loan_amount skew before:", df_treated["loan_amount"].skew().round(2), "-> after Box-Cox:", pd.Series(loan_boxcox.ravel()).skew().round(2))
print("annual_income skew before:", df_treated["annual_income"].skew().round(2), "-> after Yeo-Johnson:", pd.Series(income_yeojohnson.ravel()).skew().round(2))
```

### Construct New Features

```python
# ===== New Feature 1: Debt-to-Income Ratio =====
df_treated["debt_to_income_ratio"] = (df_treated["loan_amount"] / df_treated["annual_income"]).round(4)
df_treated[["loan_amount", "annual_income", "debt_to_income_ratio"]].head()
```

```python
# ===== New Feature 2: Average Monthly Transactions =====
df_treated["avg_monthly_transactions"] = (df_treated["transaction_count"] / 6).round(2)
df_treated[["transaction_count", "avg_monthly_transactions"]].head()
```

## Part H: Final Deliverable

```python
# ===== Build final ML-ready dataset =====
final_df = df_treated.copy()

# drop raw/intermediate columns now that encoded/engineered versions exist
final_df = final_df.drop(columns=["education_level", "gender", "join_date"])

print("Final dataset shape:", final_df.shape)
final_df.head()
```

```python
# ===== Save final cleaned & transformed dataset =====
final_df.to_csv("final_cleaned_dataset.csv", index=False)
print("Saved -> final_cleaned_dataset.csv")
```

### Report Summary

**Missing value strategies used:**
- `age` → SimpleImputer (median) | `gender` → most-frequent | `employment_type` → SimpleImputer (most-frequent)
- `annual_income`, `loan_amount`, `credit_score` → KNN Imputer (multivariate, correlated columns)
- Also demonstrated: Missing Indicator + Random Sample, MICE, Complete Case Analysis (not used in final — too much data loss)

**Outlier handling results:**
- `credit_score` → domain-clipped to valid [300, 850] range
- `annual_income`, `loan_amount` → IQR-based capping (Winsorization-style), row count preserved at 600

**Encoding methods applied:**
- Ordinal → `education_level` | Label → `gender` | One-Hot → `region`, `loan_purpose`
- Binning → `annual_income` (income_band) | Binarization → `credit_score` (good_credit_flag)
- Quantile Binning + K-Means Binning → `transaction_count`

**Scaling/transformations applied:**
- Standardization, Normalization, Min-Max, MaxAbs, Robust Scaling all demonstrated on financial columns
- Log/Sqrt/Reciprocal (FunctionTransformer) on `spending_ratio`; Box-Cox & Yeo-Johnson (PowerTransformer) on `loan_amount`/`annual_income`
- `ColumnTransformer` used to combine numeric + categorical preprocessing in one pipeline

**Newly engineered features:**
- `debt_to_income_ratio`, `avg_monthly_transactions`, `spending_to_income_ratio`
- `join_year`/`join_month`/`join_day`/`join_weekday` from `join_date`

**Final dataset readiness:** 0 missing values, outliers capped, all categoricals encoded, all numerics scaled/transformed — **ready for ML modeling** (next phase, outside this project's scope). ✅
